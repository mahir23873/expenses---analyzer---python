```python
import pandas as pd 
```

```python
df=pd.read_csv('expenses.csv')
```

```python
df
```

```python
import matplotlib.pyplot as plt
```

```python
category_totals = df.groupby("category")["amount"].sum()
```

```python
category_totals.plot(kind="bar")
plt.title("Total Spending by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.show()
```
